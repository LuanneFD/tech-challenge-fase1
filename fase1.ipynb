## Padronização de variáveis

```python
from sklearn.preprocessing import StandardScaler
# Normalizando os dados utilizando o StandardScaler
# (Padroniza as features removendo a média e escala a variância a uma unidade.
# Isso significa que para cada feature, a média seria 0, e o Desvio Padrão seria 1)
features = baseDados.drop(columns=['id', 'diagnosis']).columns
X = baseDados[features]
X = StandardScaler().fit_transform(X)
# Visualizando os dados padronizados
df_padronizado = pd.DataFrame(data=X, columns=features)
display(df_padronizado.head())
```

Depois da padronização, foi possível perceber por exemplo que a coluna com maior correlação com o diagnóstico continua sendo a concave points_worst.

# Redução de Dimensionalidade - PCA - Calcula a variância explicada acumulada

```python
import numpy as np
from sklearn.decomposition import PCA

# Calcula a variância explicada acumulada
pca = PCA()
pca.fit(df_padronizado)
variancia_cumulativa = np.cumsum(pca.explained_variance_ratio_)

# Visualização da variância explicada acumulada
plt.plot(range(1, len(variancia_cumulativa) + 1), variancia_cumulativa, marker='o')
plt.xlabel('Número de Componentes Principais')
plt.ylabel('Variância Acumulada Explicada')
plt.title('Variância Acumulada Explicada pelo PCA')
plt.show()

# Vamos definir um limiar de 80%, ou seja, queremos obter uma porcentagem de explicância sobre
# os dados de igual a 80%
limiar_de_variancia = 0.80

# Encontrar o número de componentes necessários para atingir ou ultrapassar o limiar
num_de_pca = np.argmax(variancia_cumulativa >= limiar_de_variancia) + 1

print(f"Número de Componentes para {limiar_de_variancia * 100}% da Variância: {num_de_pca}")
```

Verificando pelo gráfico é possível identificar o número de componentes ideal é o 5, portanto no bloco abaixo será aplicado o PCA para reduzir as 30 colunas para 5.

```python
# Inicializa o objeto PCA
pca = PCA(n_components=num_de_pca)
# Aplica o PCA aos dados padronizados
principal_components = pca.fit_transform(df_padronizado)
# Exibe a proporção de variância explicada
explained_variance_ratio = pca.explained_variance_ratio_
print(explained_variance_ratio)

# Pegando o número de componentes principais gerados
num_components = principal_components.shape[1]
# Gerando uma lista para cada PCA
column_names = [f'PC{i}' for i in range(1, num_components + 1)]
# Criando um novo dataframe para visualizarmos como ficaram os dados reduzidos com o PCA
pca_df = pd.DataFrame(data=principal_components, columns=column_names)
pca_df
     
```

Após rodar o PCA e verificar a proporção de variância explicada com os 5 valores: 
 0.4427 + 0.1897 + 0.0939 + 0.0660 + 0.0550 = 0.8474, temos um percentual de 84.7%.
 Portando a variância de 80% foi atendida conforme o esperado.

## Verificando normalidade após o PCA

```python
import seaborn as sns

# Criando histogramas para cada componente principal
plt.figure(figsize=(15, 4))
for i, col in enumerate(pca_df.columns):
    plt.subplot(1, len(pca_df.columns), i + 1)
    sns.histplot(pca_df[col], bins=20, kde=True)
    plt.title(f'Histograma {col}')
plt.tight_layout()
plt.show()
```

Percebi que o PC1 não gerou um gráfico com distribuição normal, a maioria dos dados está acumulada na parte esquerda que apresenta um pico. O mesmo se parece um pouco na coluna PC3.

```python
from scipy.stats import shapiro
# Verificando para cada coluna a normalidade após a redução de dimensionalidade
for column in pca_df.columns:
    stat, p_value = shapiro(pca_df[column])
    print(f'Variável: {column}, Estatística de teste: {stat}, Valor p: {p_value}')

    if p_value > 0.05:
        print(f'A variável {column} parece seguir uma distribuição normal.\n')
    else:
        print(f'A variável {column} não parece seguir uma distribuição normal.\n')
```

# Modelagem

Pensando que temos um Target definido nessa base, que seria o Diagnóstico como Maligno ou Benigno, vou usar algoritmos supervisionados como esses abaixo:

- K-Nearest Neighbors (KNN)
- Regressão Logística
- Máquinas de Vetores de Suporte (SVM)
- Árvores de Decisão

Vou testar todos eles e entender qual deles irá retornar o melhor resultado.

**Sobre a Regressão Linear**: a Regressão Linear **não foi testada** porque é feita para prever valores contínuos e não categorias como maligno/benigno.

```python
from sklearn.model_selection import train_test_split

# X = componentes principais (resultado do PCA), y = diagnóstico (1 = maligno, 0 = benigno)
X_train, X_test, y_train, y_test = train_test_split(
    pca_df,
    baseDados['diagnosis'],
    test_size=0.2,
    random_state=42,
    stratify=baseDados['diagnosis']  # mantém a mesma proporção de maligno/benigno no treino e no teste
)

print(f"Treino: {X_train.shape[0]} amostras")
print(f"Teste: {X_test.shape[0]} amostras")
```

Abaixo será feito o treinamento usando os algoritmos mencionados anteriormente e logo depois será emitido um relatório pra identificar qual deles retornou o melhor resultado. Os resultados serão avaliados usando as três métricas: **acurácia**, **recall** e **F1-score**.

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, recall_score, f1_score, classification_report

modelos = {
    'KNN': KNeighborsClassifier(),
    'Regressão Logística': LogisticRegression(random_state=42),
    'SVM': SVC(random_state=42),
    'Árvore de Decisão': DecisionTreeClassifier(random_state=42),
}

resultados = []
modelos_treinados = {}

for nome, modelo in modelos.items():
    modelo.fit(X_train, y_train)
    y_pred = modelo.predict(X_test)

    acc = accuracy_score(y_test, y_pred)
    rec = recall_score(y_test, y_pred,pos_label=1)  # recall da classe maligno (1)
    f1 = f1_score(y_test, y_pred)

    resultados.append({'Modelo': nome, 'Acurácia': acc, 'Recall': rec, 'F1-score': f1})
    modelos_treinados[nome] = modelo

    relatorio = classification_report(y_test, y_pred, target_names=['Benigno', 'Maligno'], output_dict=True)
    relatorio_df = pd.DataFrame(relatorio).transpose().loc[['Benigno', 'Maligno']]

    print(f"--- {nome} ---")
    display(relatorio_df)

df_resultados = pd.DataFrame(resultados).sort_values(by='Recall', ascending=False).reset_index(drop=True)
display(df_resultados)
```

## Análise dos resultados

Após a execução e verificação do resultado dos algoritmos, foi identificado que o modelo Regressão Logística obteve o melhor resultando avaliando todas as métricas. 
Todos os modelos resultaram em média em percentuais altos acima de 95%, isso pode mostrar que talvez o resultado esteja enviesado, acredito que seja porque o dataset está desbalanceado como foi possível observar durante o pré processamento da base com 357 casos benignos e 212 malignos.

### Escolha da métrica

- **Acurácia** Esta métrica mede o percentual total de acertos mas pode não ser o ideal nesse caso de avaliação de saúde porque esse dataset está desbalanceado.
- **Recall** (da classe maligno): O recall foi feito avaliando apenas os malignos por se tratar se um caso de saúde onde deixar passar os casos Malignos é pior porque pode atrasar o tratamento do paciente. 
- **F1-score** Esta métrica é o ideal nesse caso porque diferente do recall avalia os dois lados, o quanto o modelo acertou dos casos que eram realmente benignos ou malignos.

**Conclusão da escolha**: dado o contexto (auxílio a diagnóstico médico), a métrica f1-score é a mais relevante porém avaliando juntamente com o Recall, assim cercando os dois tipos de diagnósticos e garantindo maior precisão.

## Interpretação dos resultados utilizando o Shap.
Verificação dos resultados dos modelos regressão logística e SVM que foram os modelos que obtiveram maior pontuação olhando as 3 métricas.

```python
# SHAP para a Regressão Logística
import shap

explainer_lr = shap.LinearExplainer(modelos_treinados['Regressão Logística'], X_train)
shap_values_lr = explainer_lr(X_test)

shap.summary_plot(shap_values_lr, X_test)
```

Ao analisar o gráfico SHAP da Regressão Logística, é possível identificar que o PC1 é o componente que mais influencia a decisão do modelo. Valores altos de PC1 (em vermelho) empurram a previsão para maligno, enquanto valores baixos (em azul) empurram para benigno.
O PC2 é segundo mais importante, mas com um comportamento invertido: valores altos empurram para benigno, e valores baixos empurram para maligno.
Já os componentes PC3, PC4 e PC5 têm pouca influência na decisão do modelo, com os pontos concentrados bem próximos de zero, mostrando que contribuem pouco comparado aos dois primeiros.

De forma geral, o modelo se baseou principalmente em PC1 e PC2 para diferenciar os casos malignos dos benignos.

```python
# SHAP para o SVM

background = shap.sample(X_train, 50, random_state=42)
explainer_svm = shap.Explainer(modelos_treinados['SVM'].decision_function, background)
shap_values_svm = explainer_svm(X_test)

shap.summary_plot(shap_values_svm, X_test)
```

No gráfico SHAP do SVM, o PC1 continua sendo o componente mais importante. Valores baixos de PC1 (em azul) empurram a previsão para benigno, enquanto valores altos (em vermelho) empurram para maligno, exatamente o mesmo padrão visto na Regressão Logística.
O PC2 continua como o segundo mais relevante, também com o comportamento invertido: valores altos empurram para benigno, e valores baixos empurram para maligno.

De forma geral, o resultado do SVM confirma o que já havia sido observado na Regressão Logística: o modelo se apoia principalmente em PC1 e PC2 para diferenciar casos malignos de benignos.

## Discussão crítica: o modelo pode ser utilizado na prática?

Os modelos apresentaram desempenho alto (acurácia e recall acima de 95% na Regressão Logística e no SVM). Ainda assim, esse resultado não significa que o modelo já esteja pronto para uso clínico real, já que foi treinado com apenas 569 casos de uma única fonte, sem validação em outros hospitais ou populações.

Sim, o modelo pode ser utilizado na prática, mas apenas como uma ferramenta de apoio à decisão, nunca como o diagnóstico final. Ele poderia ser utilizado da seguinte forma:

- Sinalizando a probabilidade de malignidade para a revisão mais rápida por um médico.
- Mostrando ao médico quais características do exame (tamanho, forma, textura etc.) mais influenciaram aquela previsão, através das técnicas de explicabilidade (SHAP), para que o profissional avalie se aquilo faz sentido clinicamente antes de decidir, porém de forma bem detalhada e não como o resultado final desse estudo com as colunas PC1,PC2 e etc.
- Rodando localmente, em ambiente seguro e criptografado, sem expor dados sensíveis de pacientes pela internet, respeitando normas de proteção de dados como a LGPD.

Em todos os casos, a decisão final do diagnóstico e do tratamento deve continuar sendo do médico. O modelo funciona apenas como um segundo olhar automatizado, ajudando a priorizar casos e trazer mais informação para a decisão, nunca substituindo o julgamento clínico humano.
